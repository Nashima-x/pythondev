```python
try:
    a_str = input("割られる数を入力してください：")
    a = int(a_str)
    b_str = input("割る数を入力してください：")
    b = int(b_str)
    print(f"割られる数を入力してください：{a}")
    print(f"割る数を入力してください：{b}")

    c = a / b
    print(f"{a} ÷ {b} = {c}")

except ZeroDivisionError:
    print("エラー：0で割り算しないでください")

except ValueError:
    print(f"割られる数を入力してください：{a_str}")
    print("数値を入力してください")

except:
    print("エラー: 何らかのエラーが発生しました")

print("処理を終了します")
```
