```python
words_en = []
alpha_count = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"),0)

while True:
    words = input("英単語を入力してください：")
    print(f"英単語を入力してください：{words}")

    if words == "":
        break
    else:
        words_en.append(words)
        
        for char in words:
            for i in char:
                alpha_count[i] += 1

words_en.sort() 
print(f"入力した英単語：{words_en}")

for key,value in alpha_count.items():
    if value > 0:
        print(f"{key}が{value}個ありました")
```
